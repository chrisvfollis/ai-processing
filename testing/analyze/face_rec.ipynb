```python
import pandas as pd
```

```python
file_path = '../../files/output/faceiq_data.xlsx'
dfs = pd.read_excel(file_path, sheet_name=None)

detections_df = dfs['Detections']
artifacts_df = dfs['Pipeline Artifacts']
```

```python
# ----------------------------------------------------------------------------
# Detections:
```

```python
detections_df.describe().round(2)
```

```python
n_total = len(detections_df)

min_area = (32*32)
min_conf = 0.55
```

```python
valid_area_detections = detections_df.loc[detections_df['a'] >= min_area]
valid_conf_detections = detections_df.loc[detections_df['c'] >= min_conf]
valid_a_c_detections = valid_area_detections.loc[valid_area_detections['c'] >= min_conf]

n_valid_area = len(valid_area_detections)
n_valid_conf = len(valid_conf_detections)
n_valid_a_c = len(valid_a_c_detections)

pct_valid_a = round((n_valid_area / n_total), 2) * 100
pct_valid_conf = round((n_valid_conf / n_total), 2) * 100
pct_valid_a_c = round((n_valid_a_c / n_total), 2) * 100

print(f'\n{n_valid_area} detections ({pct_valid_a}%): area >= min area\n')
print(f'{n_valid_conf} detections ({pct_valid_conf}%): confidence >= min confidence\n')
print(f'{n_valid_a_c} detections ({pct_valid_a_c}%): sufficient area AND confidence\n')
```
