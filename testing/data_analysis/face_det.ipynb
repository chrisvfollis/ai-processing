```python
n_min_a = len(sizes_df.loc[sizes_df['a'] > (32*32)])
pcnt_min_a = round((n_min_a / len(sizes_df)), 2) * 100

print(f'{n_min_a} detections ({pcnt_min_a}% of total): area >= min area\n')

n_min_wh = len(sizes_df.loc[(sizes_df['w'] >= 32) & (sizes_df['h'] >= 32)])
pcnt_min_wh = round((n_min_wh / len(sizes_df)), 2) * 100

print(f"{n_min_wh} detections ({pcnt_min_wh}% of total): width AND height >= min side length\n")
```
